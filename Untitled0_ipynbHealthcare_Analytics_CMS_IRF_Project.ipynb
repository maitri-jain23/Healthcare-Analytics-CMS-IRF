```python
!pip install pandas numpy matplotlib seaborn scikit-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset uploaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
# Dataset dimensions
print("Dataset Shape:", df.shape)

# Column names
print("\nDataset Columns:")
print(df.columns.tolist())

# Data types
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# Check unique facility IDs
print("\nUnique Facility IDs:",
      df["CMS Certification Number (CCN)"].nunique())
```

```python
# Create a copy of original dataset
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Convert certification date
df_clean["Certification Date"] = pd.to_datetime(
    df_clean["Certification Date"], errors="coerce"
)

# Extract certification year
df_clean["Certification Year"] = (
    df_clean["Certification Date"].dt.year
)

print("Data preprocessing completed!")
print("Final Dataset Shape:", df_clean.shape)

display(df_clean.head())
```

```python
ownership_counts = df_clean["Ownership Type"].value_counts()

print("Ownership Distribution:")
print(ownership_counts)

plt.figure(figsize=(9, 5))
ownership_counts.plot(kind="bar")

plt.title("Distribution of IRF Facilities by Ownership Type")
plt.xlabel("Ownership Type")
plt.ylabel("Number of Facilities")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
state_counts = df_clean["State"].value_counts()

print("Top 10 States:")
print(state_counts.head(10))

plt.figure(figsize=(12, 6))
state_counts.head(10).plot(kind="bar")

plt.title("Top 10 States by IRF Facility Distribution")
plt.xlabel("State")
plt.ylabel("Number of Facilities")
plt.tight_layout()
plt.show()
```

```python
region_counts = df_clean["CMS Region"].value_counts().sort_index()

print("CMS Region Distribution:")
print(region_counts)

plt.figure(figsize=(10, 5))
region_counts.plot(kind="bar")

plt.title("IRF Facility Distribution Across CMS Regions")
plt.xlabel("CMS Region")
plt.ylabel("Number of Facilities")
plt.tight_layout()
plt.show()
```

```python
year_counts = df_clean["Certification Year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
year_counts.plot(kind="line", marker="o")

plt.title("IRF Facilities by Certification Year")
plt.xlabel("Certification Year")
plt.ylabel("Number of Facilities")
plt.tight_layout()
plt.show()
```

```python
state_share = df_clean["State"].value_counts(normalize=True)

HHI = (state_share ** 2).sum()

print("Geographic HHI:", round(HHI, 4))
print("HHI Scale (0-10000):", round(HHI * 10000, 2))
```

```python
ownership_share = df_clean["Ownership Type"].value_counts(normalize=True)

entropy = -(ownership_share * np.log(ownership_share)).sum()

max_entropy = np.log(len(ownership_share))

normalized_entropy = entropy / max_entropy

print("Shannon Entropy:", round(entropy, 4))
print("Normalized Ownership Diversity:",
      round(normalized_entropy, 4))
```

```python
print("========== PROJECT RESULTS ==========")

print("Total Facilities:", len(df_clean))

print("Total States:", df_clean["State"].nunique())

print("\nOwnership Distribution:")
print(ownership_counts)

print("\nTop 5 States:")
print(state_counts.head(5))

print("\nCMS Region Distribution:")
print(region_counts)

print("\nGeographic HHI:", round(HHI, 4))

print("Ownership Diversity:",
      round(normalized_entropy, 4))

print("=====================================")
```

```python
df_clean.to_csv("Cleaned_CMS_IRF_Dataset.csv", index=False)

summary = pd.DataFrame({
    "Metric": [
        "Total Facilities",
        "Total States",
        "Geographic HHI",
        "Normalized Ownership Diversity"
    ],
    "Value": [
        len(df_clean),
        df_clean["State"].nunique(),
        HHI,
        normalized_entropy
    ]
})

summary.to_csv("FDOP_Analysis_Results.csv", index=False)

print("Project files saved successfully!")
```
